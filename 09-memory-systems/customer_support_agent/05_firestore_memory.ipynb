# Topic 5 — Firestore: The Durable Customer Profile

This is where topic 2's long-term memory actually belongs in production. Uses `support_customer_profiles` — a different collection name than the original module's `long_term_memory`, so both can coexist in the same project.

```python
from setup import firestore_client
from google.cloud.firestore_v1 import ArrayUnion

profiles = firestore_client.collection("support_customer_profiles")
```

### Create the profile

```python
profiles.document("cust_042").set({
    "plan_tier": "Enterprise",
    "preferred_contact": "email",
    "known_issues": ["Billing issue resolved Jan 5"],
}, merge=True)

print(profiles.document("cust_042").get().to_dict())
```

### A new issue comes in — append without overwriting the rest

```python
profiles.document("cust_042").update({
    "known_issues": ArrayUnion(["App crashes on opening settings - reported today"]),
})

print(profiles.document("cust_042").get().to_dict())
```
