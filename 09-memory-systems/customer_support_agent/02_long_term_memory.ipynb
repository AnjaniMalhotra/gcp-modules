# Topic 2 — Long-Term Memory: Remembering a Returning Customer

A support agent that asks "what plan are you on?" every single time a customer returns is worse than a human agent who glances at a CRM first. This notebook uses a local file to stand in for durable storage — topic 5 (Firestore) makes it production-real.

```python
import json
from pathlib import Path

MEMORY_FILE = Path("support_customer_facts.json")

def remember_fact(customer_id: str, fact: str):
    facts = json.loads(MEMORY_FILE.read_text()) if MEMORY_FILE.exists() else {}
    facts.setdefault(customer_id, []).append(fact)
    MEMORY_FILE.write_text(json.dumps(facts, indent=2))

def get_facts(customer_id: str) -> list:
    if not MEMORY_FILE.exists():
        return []
    return json.loads(MEMORY_FILE.read_text()).get(customer_id, [])
```

### The customer's first visit

```python
remember_fact("cust_042", "On the Enterprise plan")
remember_fact("cust_042", "Had a billing issue resolved on Jan 5")
print("Saved.")
```

### The customer returns next week — simulate a fresh process, read from disk

```python
facts = get_facts("cust_042")
print(f"SupportBot already knows about cust_042: {facts}")
```
