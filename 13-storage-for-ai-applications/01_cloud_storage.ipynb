# Topic 1 — Cloud Storage: Student Documents Vault

Object storage holds whole files. This notebook demos the basic CRUD operations: create (upload), read, update (full overwrite), and delete — plus a signed URL for temporary access.

```python
from setup import GCS_BUCKET_NAME, PROJECT_ID
from google.cloud import storage
import datetime

client = storage.Client(project=PROJECT_ID)
bucket = client.bucket(GCS_BUCKET_NAME)
```

### Create — upload a "report card"

```python
blob = bucket.blob("students/alex_report_card.txt")
blob.upload_from_string("Student: Alex Chen\nGrade: A\nAttendance: 95%")
print("Uploaded.")
```

### Read — list the bucket, download the file

```python
print("Files in bucket:")
for b in bucket.list_blobs(prefix="students/"):
    print(" -", b.name)

print("\nContent:")
print(blob.download_as_text())
```

### Update — a full overwrite, not a partial edit

There's no "just change the attendance line" — the whole object gets replaced.

```python
blob.upload_from_string("Student: Alex Chen\nGrade: A+\nAttendance: 97%")
print("Overwritten:")
print(blob.download_as_text())
```

### Signed URL — temporary access without sharing credentials

```python
signed_url = blob.generate_signed_url(
    version="v4",
    expiration=datetime.timedelta(minutes=15),
    method="GET",
)
print("Temporary link (valid 15 minutes):")
print(signed_url)
```

### Delete

```python
blob.delete()
print("Deleted.")
```
