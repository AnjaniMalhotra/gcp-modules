# Topic 2 — Firestore: Student Profiles

Flexible, nested documents — a different shape of data than topic 1's whole files or topic 3's rigid relational rows.

```python
from setup import PROJECT_ID
from google.cloud import firestore
from google.cloud.firestore_v1 import ArrayUnion

db = firestore.Client(project=PROJECT_ID)
students = db.collection("students")
```

### Create — a profile with nested + array fields

```python
students.document("alex").set({
    "name": "Alex Chen",
    "grade": 10,
    "contact": {"email": "alex@example.com", "phone": "555-0100"},
    "extracurriculars": ["chess club", "robotics"],
})

students.document("priya").set({
    "name": "Priya Nair",
    "grade": 11,
    "contact": {"email": "priya@example.com", "phone": "555-0101"},
    "extracurriculars": ["debate team"],
})

print("Created 2 student profiles.")
```

### Read

```python
doc = students.document("alex").get()
print(doc.to_dict())
```

### Query — every grade-10 student

```python
print("Grade 10 students:")
for doc in students.where("grade", "==", 10).stream():
    print(" -", doc.to_dict()["name"])
```

### Update — append to an array without overwriting the rest of the document

```python
students.document("alex").update({"extracurriculars": ArrayUnion(["debate team"])})
print(students.document("alex").get().to_dict())
```

### Delete

```python
students.document("priya").delete()
print("Deleted priya's profile.")
```
